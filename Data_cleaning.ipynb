```python
import pandas as pd
```

```python
#load data/excel

df = pd.read_excel("sales_raw_500.xlsx")
```

```python
df
```

```python
#remove duplicate rows

df = df.drop_duplicates()
```

```python
# rows after duplicte removed

df
```

```python
#handling missing data in columns in Order_Date

df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors= 'coerce')
```

```python
#drop the row after conversion of columen formatted

df = df.dropna(subset=['Order_Date'])
```

```python
df
```

```python
# adding 3 columns for profits, extracting month & year to new columns

df['Profit'] = df['Sales'] -df ['Cost']
df['Year'] = df['Order_Date'].dt.year
df['Month'] = df['Order_Date'].dt.month
```

```python
df
```

```python
#export the clear to a csv file in local, index =false for not start with 0 row

df.to_csv('Sale_clearn.csv', index = False)
```
